```sql
-- SLA performance by region and month

select
region, 
department,
date_trunc('month', created_at)                                                          as month,
count(*)                                                                                 as total_tickets,
sum(case when minutes_to_first_response <=60 then 1 else 0 end)                          as within_sla,

round(
    sum(case when minutes_to_first_response <=60 then 1 else 0 end) * 100.0 /count(*), 2 
)                                                                                          as sla_pct,
round(avg(minutes_to_first_response), 1)                                                   as avg_first_response_min,
round(avg(hours_to_resolution), 1)                                                         as avg_resolution_hrs,
sum(is_open)                                                                               as open_tickets

from gold.tickets_clean
group by 1, 2, 3
order by 1 , 2;
```
