```sql
create or replace table gold.tickets_clean as 

select 
ticket_id,
customer_id,
agent_id,
upper(trim(region))                   as region,
upper(trim(department))               as department,
cast(created_at as timestamp)         as created_at,
cast(first_response_at as timestamp)  as first_response_at,
cast(resolved_at as timestamp)        as resolved_at,
case
    when priority in ('high', 'urgent') then 'High'
    when priority = 'medium'  then           'Medium'
    else                                     'Low'
    end      as priority_bucket,
    Datediff(minute, created_at, first_response_at)     as minutes_to_first_response,
    datediff(hour, created_at, resolved_at)             as hours_to_resolution,
    case when resolved_at is null then 1 else 0 end     as is_open
from raw.support_tickets
where created_at >= '2023-01-01'
and ticket_id is not null;

-- select * from gold.tickets_clean;
```
