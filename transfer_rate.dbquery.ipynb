```sql
-- Transfer rate : tickets escalated vs total , by department 

select 
t.department,
count(distinct t.ticket_id)                     as total_tickets,
count(distinct e.ticket_id)                     as transferred_tickets,
round(count(distinct e.ticket_id) * 100.0
        / count(distinct t.ticket_id), 2)      as transfer_rate_pct

from gold.tickets_clean t 
left join raw.escalations e  
on t.ticket_id = e.ticket_id   
group by 1 
order by transfer_rate_pct desc;
```
